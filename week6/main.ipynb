## 문제

디랩이는 여러 도시에 디랩캠퍼스를 두고 있다.
이 도시는 1번부터 V번까지 번호가 붙어 있으며, 일부 도시는 버스로 연결되어 있다.
모든 버스는 일방통행이며, 비용이 정해져 있다.

디랩이는 특정한 시작 도시 K에서 출발해, 나머지 모든 도시에 도달하는 데 필요한 최소 비용을 알고 싶다.
만약 어떤 도시는 도달할 수 없다면, 디랩이는 그곳에는 갈 수 없으므로 "INF"라고 표시하기로 했다.

---

### 입력

- 첫째 줄에 도시(정점)의 개수 V와 버스(간선)의 개수 E가 주어진다. (1 ≤ V ≤ 20,000, 1 ≤ E ≤ 300,000)
- 둘째 줄에 시작 도시 K (1 ≤ K ≤ V)가 주어진다.
- 셋째 줄부터 E개의 줄에 걸쳐 버스 정보가 주어진다.
  - 각 줄은 세 정수 u, v, w로 이루어지며, u번 도시에서 v번 도시로 가는 비용이 w임을 의미한다.
  - u와 v는 서로 다르며 w는 10 이하의 자연수이다.
  - 서로 다른 두 정점 사이에 여러 개의 간선이 존재할 수도 있음에 유의한다.

---

### 출력

- 각 줄에 시작 도시 K에서 해당 도시까지 가는 최소 비용을 출력한다.
- 도달할 수 없는 경우에는 "INF"를 출력한다.

---

- **시간제한**: 1초
- **메모리제한**: 256MB

```python
import heapq

def dijkstra(n, graph, start):
    INF = 1_000_000_000
    dist = [INF] * (n + 1)
    dist[start] = 0
    pq = [(0, start)]
    
    while pq:
        cost, u = heapq.heappop(pq)
        if dist[u] < cost:
            continue
        for v, w in graph[u]:
            new_cost = cost + w
            if dist[v] > new_cost:
                dist[v] = new_cost
                heapq.heappush(pq, (new_cost, v))
    
    return dist

V, E = map(int, input().split())
K = int(input())
graph = [[] for _ in range(V + 1)]

for _ in range(E):
    u, v, w = map(int, input().split())
    graph[u].append((v, w))

dist = dijkstra(V, graph, K)

INF = 1_000_000_000
for i in range(1, V + 1):
    if dist[i] >= INF:
        print("INF")
    else:
        print(dist[i])
```

## 문제

디랩이는 여러 도시를 순회하며 개발 워크숍을 열어야 한다. 각 도시는 일방통행 버스로 연결되어 있고, 버스마다 요금이 다르다.
디랩이는 특정 출발 도시에서 목표 도시에 도착하려고 한다. 디랩이는 총 이동 비용이 최소가 되도록 이동하려고 하며, 그때 실제로 어떤 도시들을 어떤 순서로 지나가면 되는지도 알고 싶다.

디랩이를 도와 최소 이동 비용과, 그 비용을 달성하는 경로 하나를 구하라.

단, 모든 입력 데이터는 최단 경로가 정확히 하나만 존재하도록 주어진다. 따라서 답은 항상 유일하다.

---

### 입력

- 첫째 줄에 도시의 개수 N이 주어진다. (1 ≤ N ≤ 1,000)
- 둘째 줄에 버스의 개수 M이 주어진다. (1 ≤ M ≤ 100,000)
- 셋째 줄부터 M개의 줄에는 버스 정보가 주어진다.
  - 각 줄은 세 정수 u, v, w로, u번 도시에서 v번 도시로 가는 버스의 비용이 w임을 의미한다. (0 ≤ w ≤ 100,000)
- 마지막 줄에는 출발 도시와 도착 도시가 공백으로 주어진다.

---

### 출력

- 첫째 줄에 출발 도시에서 도착 도시까지의 최소 비용을 출력한다.
- 둘째 줄에 최소 비용을 달성하는 경로에 포함된 도시 수를 출력한다.
- 셋째 줄에 그 경로를 이루는 도시들을 순서대로 공백으로 구분하여 출력한다.

---

- **시간제한**: 1초
- **메모리제한**: 256MB

```python
import heapq

def dijkstra(n, graph, start):
    INF = 1_000_000_000
    dist = [INF] * (n + 1)
    dist[start] = 0
    pq = [(0, start)]
    parent = [-1] * (n + 1)
    parent[start] = -1
    
    while pq:
        cost, u = heapq.heappop(pq)
        if dist[u] < cost:
            continue
        for v, w in graph[u]:
            new_cost = cost + w
            if dist[v] > new_cost:
                dist[v] = new_cost
                parent[v] = u
                heapq.heappush(pq, (new_cost, v))
    
    return dist, parent

N = int(input())
M = int(input())
graph = [[] for _ in range(N + 1)]

for _ in range(M):
    u, v, w = map(int, input().split())
    graph[u].append((v, w))

start, end = map(int, input().split())
dist, parent = dijkstra(N, graph, start)

# 경로 추적
path = []
node = end
while node != -1:
    path.append(node)
    node = parent[node]

path.reverse()

print(dist[end])
print(len(path))
print(' '.join(map(str, path)))
```
