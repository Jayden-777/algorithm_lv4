```python
import heapq
N = int(input())
h = []
number = 0
total = 0
nums = [int(input()) for _ in range(N)]
for i in nums:
    heapq.heappush(h,i)
while len(h) > 1:
    card1 = heapq.heappop(h)
    card2 = heapq.heappop(h)
    number = card1 + card2
    heapq.heappush(h, number)
    total += number
print(total)
```
