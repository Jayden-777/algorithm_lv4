## 문제

디랩이는 전국 코딩 캠프를 돌며 학생들을 만나야 합니다.

각 도시는 버스로 연결되어 있고, 노선마다 비용이 다릅니다. 디랩이는 출발 도시에서 도착 도시까지 드는 **최소 비용**을 알고 싶습니다. (시간이 아닌 비용 최소화이며, 같은 도시 쌍 사이에 여러 노선이 있을 수도 있습니다.)

디랩이는 지갑 사정을 생각해, 가장 저렴한 경로를 찾아 떠납니다. 당신은 디랩이를 도와 최소 비용을 계산해 주세요!

---

### 입력

- 첫째 줄에 도시의 개수 N ($1 \le N \le 1\,000$)
- 둘째 줄에 버스의 개수 M ($1 \le M \le 100\,000$)
- 셋째 줄부터 M개의 줄에 걸쳐 버스 정보가 주어집니다. 각 줄은 `u v w` 형식이며, u번 도시에서 v번 도시로 가는 비용이 w입니다.
- 도시는 **방향성이 있는** 버스로 연결됩니다 ($u \to v$). 같은 $(u, v)$ 쌍에 대해 여러 개의 노선이 주어질 수 있습니다.
- 비용 w는 양의 정수이며 $100\,000$ 이하입니다.
- 마지막 줄에 출발 도시와 도착 도시가 공백으로 주어집니다.

---

### 출력

- 출발 도시에서 도착 도시까지 필요한 최소 비용을 한 줄에 출력합니다.
- 경로가 존재함이 보장됩니다.

---

- **시간제한**: 0.5초
- **메모리제한**: 128MB

```python
# 알고리즘: 다익스트라 (인접 리스트, 배열 기반)

def dijkstra(n, graph, start, end):
    INF = 1_000_000_000
    dist = [INF] * (n + 1)
    visited = [False] * (n + 1)
    dist[start] = 0
    for _ in range(n):
        min_dist = INF
        u = -1
        for i in range(1, n + 1):
            if not visited[i] and dist[i] < min_dist:
                min_dist = dist[i]
                u = i
        if u == -1:
            break
        visited[u] = True
        for v, cost in graph[u]:
            if dist[v] > dist[u] + cost:
                dist[v] = dist[u] + cost
    return dist[end]


N = int(input())
M = int(input())
graph = [[] for _ in range(N + 1)]
for _ in range(M):
    u, v, w = map(int, input().split())
    graph[u].append((v, w))
start, end = map(int, input().split())
print(dijkstra(N, graph, start, end))
```

## 문제

디랩이는 오늘도 전국을 돌며 학생들을 만나러 간다. 1번 도시에서 출발해 N번 도시에 도착해야 하는데, 반드시 거쳐야 하는 두 곳이 있다. 그곳은 디랩이의 소중한 연구소(v1)와 캠프장(v2)이다.

디랩이는 어떤 순서로 두 장소를 들러도 상관없다. 즉, $1 \to v_1 \to v_2 \to N$ 이거나 $1 \to v_2 \to v_1 \to N$ 으로 이동할 수 있다면 된다. 목표는 가장 저렴한 경로로 이동하는 것이다.

디랩이를 도와서 최소 비용을 구해 주자. 만약 어떤 경우에도 두 장소를 모두 거쳐서 N번 도시에 도착할 수 없다면 **-1**을 출력해야 한다.

---

### 입력

- 첫째 줄에 도시의 개수 N과 도로의 개수 M이 주어진다. ($2 \le N \le 800$, $0 \le M \le 200\,000$)
- 둘째 줄부터 M개의 줄에 걸쳐 세 정수 a, b, c가 주어진다. a번 도시와 b번 도시가 **양방향** 도로로 연결되어 있으며, 비용이 c이다. ($1 \le c \le 1\,000$)
- 임의의 두 정점 u, v 사이에는 간선이 최대 1개이다.
- 마지막 줄에는 반드시 거쳐야 하는 두 정점 v1, v2가 주어진다.
- $v_1 \ne v_2$, $v_1 \ne N$, $v_2 \ne 1$

---

### 출력

- 1번 도시에서 N번 도시로 이동하는 최소 비용을 출력한다.
- 만약 경로가 없다면 -1을 출력한다.

---

- **시간제한**: 1초
- **메모리제한**: 256MB

```python
# 알고리즘: 다익스트라 3회 (1, v1, v2에서) + 두 순서 중 최소

def dijkstra(n, graph, start):
    INF = 1_000_000_000
    dist = [INF] * (n + 1)
    visited = [False] * (n + 1)
    dist[start] = 0
    for _ in range(n):
        min_dist = INF
        u = -1
        for i in range(1, n + 1):
            if not visited[i] and dist[i] < min_dist:
                min_dist = dist[i]
                u = i
        if u == -1:
            break
        visited[u] = True
        for v, cost in graph[u]:
            if dist[v] > dist[u] + cost:
                dist[v] = dist[u] + cost
    return dist


INF = 1_000_000_000
N, M = map(int, input().split())
graph = [[] for _ in range(N + 1)]

for _ in range(M):
    a, b, c = map(int, input().split())
    graph[a].append((b, c))
    graph[b].append((a, c))

v1, v2 = map(int, input().split())

dist_1 = dijkstra(N, graph, 1)
dist_v1 = dijkstra(N, graph, v1)
dist_v2 = dijkstra(N, graph, v2)

path1 = dist_1[v1] + dist_v1[v2] + dist_v2[N]
path2 = dist_1[v2] + dist_v2[v1] + dist_v1[N]

ans = min(path1, path2)
if ans >= INF:
    print(-1)
else:
    print(ans)
```
