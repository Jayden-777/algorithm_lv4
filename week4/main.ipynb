## 문제

디랩이는 요즘 체력도 기르고 머리도 식힐 겸, 학원 근처 마을들을 돌며 운동하는 습관을 들이기로 했습니다. 마을들 사이에는 일방통행 도로들이 놓여 있어서, 정해진 방향으로만 이동할 수 있어요. 마을에는 1번부터 V번까지 번호가 붙어 있고, 각 도로는 거리도 다 다릅니다.

디랩이는 운동을 시작한 마을로 다시 돌아와야만 운동을 마쳤다고 생각합니다. 그래서 도로를 따라 어떤 마을에서 출발해서 다시 그 마을로 돌아오는 순환 경로를 찾으려고 합니다. 하지만 디랩이는 운동을 오래 하는 걸 정말 싫어해서, 그 순환 경로의 총 거리가 가능한 한 짧았으면 좋겠다고 합니다.

디랩이를 도와서, 도로를 따라 다시 시작점으로 돌아올 수 있는 경로 중에서 가장 짧은 거리를 찾아주세요.  
단, 어떤 두 마을 사이를 서로 오가는 것도 순환 경로에 포함된다는 점을 꼭 기억하세요.

---

### 입력

- 첫째 줄에 마을의 수 V와 도로의 수 E가 주어집니다. (2 이상 400 이하)
- 다음 E줄에는 각 도로의 정보가 주어집니다.
- 각 줄에는 세 개의 정수 a, b, c가 주어지며, 이는 a번 마을에서 b번 마을로 가는 도로가 있고 그 거리가 c라는 뜻입니다.
- 모든 거리는 10,000 이하의 자연수이며, 같은 도로가 여러 번 주어지지 않습니다.

---

### 출력

- 출발한 마을로 다시 돌아오는 경로 중에서 총 거리가 가장 짧은 것을 찾아 그 길이를 출력합니다.
- 만약 그런 경로가 없다면 -1을 출력합니다.

---

- **시간제한**: 2초
- **메모리제한**: 192MB

## 한국정보올림피아드 2011 지역본선 초등부 4번

---

### 문제

1번부터 N번까지 번호가 붙여져 있는 학생들에 대하여 두 학생끼리 키를 비교한 결과의 일부가 주어져 있다. 단, N명의 학생들의 키는 모두 다르다고 가정한다.  
예를 들어, 6명의 학생들에 대하여 6번만 키를 비교하였고, 그 결과가 다음과 같다고 하자.

- 1번 학생의 키 < 5번 학생의 키  
- 3번 학생의 키 < 4번 학생의 키  
- 5번 학생의 키 < 4번 학생의 키  
- 4번 학생의 키 < 2번 학생의 키  
- 4번 학생의 키 < 6번 학생의 키  
- 5번 학생의 키 < 2번 학생의 키  

이 비교 결과로부터 모든 학생 중에서 키가 가장 작은 학생부터 자신이 몇 번째인지 알 수 있는 학생들도 있고 그렇지 못한 학생들도 있다는 사실을 아래처럼 그림을 그려 쉽게 확인할 수 있다.  
a번 학생의 키가 b번 학생의 키보다 작다면, a에서 b로 화살표를 그려서 표현하였다.  

1번은 5번보다 키가 작고, 5번은 4번보다 작기 때문에, 1번은 4번보다 작게 된다.  
그러면 1번, 3번, 5번은 모두 4번보다 작게 된다.  
또한 4번은 2번과 6번보다 작기 때문에, 4번 학생은 자기보다 작은 학생이 3명이 있고, 자기보다 큰 학생이 2명이 있게 되어 자신의 키가 몇 번째인지 정확히 알 수 있다.  
그러나 4번을 제외한 학생들은 자신의 키가 몇 번째인지 알 수 없다.

학생들의 키를 비교한 결과가 주어질 때, 자신의 키가 몇 번째인지 알 수 있는 학생들이 모두 몇 명인지 계산하여 출력하는 프로그램을 작성하시오.

---

### 입력

- 첫째 줄에 학생들의 수 N (2 ≤ N ≤ 500)과 두 학생 키를 비교한 횟수 M (0 ≤ M ≤ N(N-1)/2)이 주어진다.
- 다음 M개의 각 줄에는 두 학생의 키를 비교한 결과를 나타내는 N보다 작거나 같은 서로 다른 양의 정수 a와 b가 주어진다.  
  이는 번호가 a인 학생이 번호가 b인 학생보다 키가 작은 것을 의미한다.

---

### 출력

- 자신이 키가 몇 번째인지 알 수 있는 학생이 모두 몇 명인지를 출력한다.

---

### 시간제한

- 1초

---

### 메모리제한

- 128MB

```python

INF = 1_000_000_000

N,M = map(int,input().split())

dist = [[INF] * (N+1) for _ in range(N+1)]
num = []

for _ in range(N+1):
    num.append(0)

total = 0 

for _ in range(M):
    a,b =map(int,input().split())
    dist[a][b] = 1

for k in range(1,N +1):
    for i in range(1,N +1):
        for j in range(1, N+1):
            dist[i][j] = min(dist[i][j],dist[i][k]+dist[k][j])

for i in range(1,N +1):
    for j in range(1,N +1):
        if dist[i][j] < INF or dist[j][i] < INF:
            num[i] += 1

for k in range(1,N+1):
    if num[k] == N-1:
        total += 1

print(total)
```
